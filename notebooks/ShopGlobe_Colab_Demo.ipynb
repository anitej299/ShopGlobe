## Interactive semantic product search
Every recommendation gives evidence. Explicit feature, RAM, and price requirements are hard constraints.


```python
import ipywidgets as widgets
from IPython.display import display, clear_output
from src.semantic_search import search as semantic_search
query_box = widgets.Text(value='Waterproof running shoes for flat feet under 5000', description='Query:', layout=widgets.Layout(width='85%'))
search_button = widgets.Button(description='Search', button_style='info')
search_output = widgets.Output()
def run_search(_):
    with search_output:
        clear_output()
        try:
            rows, intent = semantic_search(query_box.value, products, graph, 8)
            print('Parsed intent:', intent)
            for row in rows:
                p = row['product']
                print(f"{p['name']} | ₹{p['price']:,.0f} | {row['score']}% | {'; '.join(row['explanation'])}")
            if not rows: print('No eligible products')
        except ValueError as exc: print(exc)
search_button.on_click(run_search)
display(query_box, search_button, search_output)
run_search(None)
```

## Focused semantic relationship visualization


```python
import networkx as nx
import matplotlib.pyplot as plt
from src.ontology_builder import SG
p = products[0]
subject = SG['product/'+p['id']]
view = nx.DiGraph()
for _, predicate, obj in graph.triples((subject,None,None)):
    if predicate in (SG.belongsToCategory,SG.manufacturedBy,SG.hasFeature,SG.suitableForActivity):
        view.add_edge(p['name'], str(obj).rsplit('/',1)[-1].replace('_',' '), label=str(predicate).rsplit('/',1)[-1])
plt.figure(figsize=(11,6))
position=nx.spring_layout(view,seed=7)
nx.draw_networkx(view,position,node_color='#bdeae7',node_size=2200,font_size=8,arrows=True)
nx.draw_networkx_edge_labels(view,position,edge_labels=nx.get_edge_attributes(view,'label'),font_size=7)
plt.axis('off');plt.show()
```
